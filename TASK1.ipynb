```python
def max_profit(n):
    # (Code, Build_time, Revenue per time unit)
    buildings = [
        ('T', 5, 1500),   # Theatre
        ('P', 4, 1000),   # Pub
        ('C', 10, 2000)   # Commercial Park
    ]

    dp = [-10**18] * (n + 1)
    parent = [None] * (n + 1)

    dp[0] = 0

    # Dynamic Programming
    for t in range(n + 1):
        if dp[t] < -1e17:
            continue
        for code, btime, revenue in buildings:
            finish = t + btime
            if finish <= n:
                gain = (n - finish) * revenue
                if dp[finish] < dp[t] + gain:
                    dp[finish] = dp[t] + gain
                    parent[finish] = (t, code)

    # Best finish time
    best_time = max(range(n + 1), key=lambda x: dp[x])
    best_earnings = dp[best_time]

    # Backtracking to count buildings
    counts = {'T': 0, 'P': 0, 'C': 0}
    t = best_time

    while t != 0 and parent[t] is not None:
        prev, code = parent[t]
        counts[code] += 1
        t = prev

    return counts, best_earnings


# -----------------------
# TEST CASES
# -----------------------
tests = [7, 8, 13]

for n in tests:
    counts, earnings = max_profit(n)
    print("Time Unit:", n)
    print(f"T: {counts['T']}  P: {counts['P']}  C: {counts['C']}")
    print("Earnings: $", int(earnings))
    print()
```

```python
%%writefile app.py
import streamlit as st

st.set_page_config(page_title="Max Profit Builder", page_icon="🏗️", layout="centered")

st.title("🏗️ Max Profit Property Builder")
st.write("Enter total available time units to get the best combination of properties.")

# --------------------------
# LOGIC FUNCTION
# --------------------------
def max_profit(n):
    buildings = [
        ('T', 5, 1500),  # Theatre
        ('P', 4, 1000),  # Pub
        ('C', 10, 2000)  # Commercial Park
    ]

    dp = [-10**18] * (n + 1)
    parent = [None] * (n + 1)
    dp[0] = 0

    for t in range(n + 1):
        if dp[t] < -1e17:
            continue
        for code, btime, revenue in buildings:
            finish = t + btime
            if finish <= n:
                gain = (n - finish) * revenue
                if dp[finish] < dp[t] + gain:
                    dp[finish] = dp[t] + gain
                    parent[finish] = (t, code)

    best_time = max(range(n + 1), key=lambda x: dp[x])
    best_earnings = dp[best_time]

    counts = {"T": 0, "P": 0, "C": 0}
    t = best_time
    while t != 0 and parent[t] is not None:
        prev, code = parent[t]
        counts[code] += 1
        t = prev

    return counts, best_earnings


# --------------------------
# USER INPUT
# --------------------------
time_input = st.number_input("⏱️ Enter Total Time Units", min_value=1, max_value=500, value=13)

if st.button("🔍 Calculate Best Combination"):
    counts, earnings = max_profit(int(time_input))

    st.subheader("📌 Optimal Solution")
    st.write(f"*Time Units:* {int(time_input)}")

    col1, col2, col3 = st.columns(3)
    col1.metric("Theatres (T)", counts["T"])
    col2.metric("Pubs (P)", counts["P"])
    col3.metric("Commercial Parks (C)", counts["C"])

    st.subheader("💰 Maximum Earnings")
    st.success(f"*${int(earnings):,}*")

    st.divider()

    st.write("### 📘 Explanation")
    st.write("""
    - Each building generates income per time unit after construction.
    - Buildings cannot be constructed in parallel.
    - Dynamic Programming is used to compute optimal profit.
    - Backtracking finds the exact building combination.
    """)

st.write("---")
st.caption("Made with ❤️ using Streamlit")
```
